```python
def calculate_label_statistics(df, label="Pathogenic"):
    
    filtered_df = df[df['Predicted_Label_Display'] == label]
    number_of_variants = filtered_df.shape[0]
    number_of_mutated_protien_seq = filtered_df[filtered_df['normal_seq'] == False].shape[0]
    number_of_normal_protien_seq = filtered_df[filtered_df['normal_seq'] == True].shape[0]
    mutated_percentage_to_total = (number_of_mutated_protien_seq / number_of_variants * 100) if number_of_variants > 0 else 0
    average_confidence = filtered_df['Prediction_Confidence'].mean()
    average_confidence_normal = filtered_df[filtered_df['normal_seq'] == True]['Prediction_Confidence'].mean()
    average_confidence_mutated = filtered_df[filtered_df['normal_seq'] == False]['Prediction_Confidence'].mean()

    return {
        "number_of_variants": number_of_variants,
        "number_of_mutated_protien_seq": number_of_mutated_protien_seq,
        "number_of_normal_protien_seq": number_of_normal_protien_seq,
        "mutated_percentage_to_total": mutated_percentage_to_total,
        "average_confidence": average_confidence,
        "average_confidence_normal": average_confidence_normal,
        "average_confidence_mutated": average_confidence_mutated
    }


pathogenic_old_model_stats = calculate_label_statistics(model, "Pathogenic")
likely_pathogenic_old_model_stats = calculate_label_statistics(model, "Likely Pathogenic")
benign_old_model_stats = calculate_label_statistics(model, "Benign")
likely_benign_old_model_stats = calculate_label_statistics(model, "Likely Benign")

print("pathogenic old model stats")
for stat in pathogenic_old_model_stats:
    print(f"Old Model - {stat}: {pathogenic_old_model_stats[stat]}")
print("=" * 50)
print("likely pathogenic old model stats")
for stat in likely_pathogenic_old_model_stats:
    print(f"Likely Pathogenic Old Model - {stat}: {likely_pathogenic_old_model_stats[stat]}")
print("=" * 50)

print("benign old model stats")
for stat in benign_old_model_stats:
    print(f"Benign Old Model - {stat}: {benign_old_model_stats[stat]}")
print("=" * 50)
print("likely benign old model stats")
for stat in likely_benign_old_model_stats:
    print(f"Likely Benign Old Model - {stat}: {likely_benign_old_model_stats[stat]}")   
```
