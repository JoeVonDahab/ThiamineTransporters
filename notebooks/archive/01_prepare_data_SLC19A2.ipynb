### Converting the txt to DataFrame

```python
import pandas as pd

# Step 1: Define the file path
file_name = "SLC19A2_dSNP.txt"

# Step 3: Parse the file into a DataFrame (assuming it's tab-separated)
df = pd.read_csv(file_name, sep="\t")
print("\nFirst 5 rows of the DataFrame:")
print(df.head())
print(f'length of the dataset: {len(df)}')


# Step 4: Save the DataFrame as a CSV file
#csv_file_name = "csv_files/parsed_file.csv"
#df.to_csv(csv_file_name, index=False)
#print(f"\nData saved as '{csv_file_name}'.")

```

```python
# Count labeled and unlabeled data
labeled_count = df['clinical_significance'].notna().sum()
unlabeled_count = df['clinical_significance'].isna().sum()

print(f"Labeled data (non-NaN clinical_significance): {labeled_count}")
print(f"Unlabeled data (NaN clinical_significance): {unlabeled_count}")
print(f"Total rows: {len(df)}")
print(f"Percentage labeled: {labeled_count/len(df)*100:.2f}%")
print(f"Percentage unlabeled: {unlabeled_count/len(df)*100:.2f}%")

# Show the distribution of clinical significance values for labeled data
print(f"\nDistribution of clinical significance values:")
print(df['clinical_significance'].value_counts(dropna=False))
```

```python

# Remove rows that repeat the column header.
# We'll assume that any row where the first column's value equals the header name should be dropped.
header_name = df.columns[0]
df_clean = df[df[header_name] != header_name].copy()

print("\nFirst 5 rows after cleaning:")
print(df_clean.head())

# Save the cleaned DataFrame as a CSV file
# csv_file_name = "csv_files/parsed_file_cleaned.csv"
#df_clean.to_csv(csv_file_name, index=False)
# print(f"\nCleaned data saved as '{csv_file_name}'.")

```

```python
df_var = df_clean.copy()

# The variation format is like 'A>G', so we split at '>' and take the first part
df_var['original'] = df_var['variation'].str.split('>', expand=True)[0]
# Get the mutant part (after '>')
df_var['mutant'] = df_var['variation'].str.split('>', expand=True)[1]

df_var.shape
```

```python
# Separate rows with multiple mutants (e.g., G>A,C)
expanded_df = []

for index, row in df_var.iterrows():
    # Check if the mutant column contains a comma (multiple mutations)
    if ',' in str(row['mutant']):
        # Split the mutant values by comma
        mutants = row['mutant'].split(',')
        
        # Create a new row for each mutant
        for mutant in mutants:
            new_row = row.copy()
            new_row['mutant'] = mutant
            expanded_df.append(new_row)
    else:
        # If there's only one mutant, keep the row as is
        expanded_df.append(row)

# Create a new DataFrame from the expanded list
expanded_df = pd.DataFrame(expanded_df)

# Reset the index of the new DataFrame
expanded_df.reset_index(drop=True, inplace=True)

# Display the first few rows to verify the expansion
expanded_df.shape

# Export the expanded DataFrame to a CSV file
#expanded_df.to_csv('csv_files/expanded_variants.csv', index=False)
# print(f"Exported {len(expanded_df)} rows to expanded_variants.csv")
```

```python
# Create a DataFrame with normal versions for each unique SNP ID
normal_variants = []

# Get unique SNP IDs
unique_snp_ids = expanded_df['snp_id'].unique()

for snp_id in unique_snp_ids:
    # Get the first row for this SNP ID as a template
    template_row = expanded_df[expanded_df['snp_id'] == snp_id].iloc[0].copy()
    
    # Create a normal version of this variant
    normal_row = template_row.copy()
    normal_row['clinical_significance'] = 'normal'
    normal_row['mutant'] = normal_row['original']  # Set mutant to be the same as original
    
    # Add to our list of normal variants
    normal_variants.append(normal_row)

# Create a DataFrame from the normal variants
normal_df = pd.DataFrame(normal_variants)

# Combine the expanded DataFrame with the normal variants
combined_df = pd.concat([expanded_df, normal_df], ignore_index=True)

# Reset the index
combined_df.reset_index(drop=True, inplace=True)

# Display the shape of the combined DataFrame
print(f"Original expanded DataFrame shape: {expanded_df.shape}")
print(f"Normal variants added: {len(normal_variants)}")
print(f"Combined DataFrame shape: {combined_df.shape}")

# Export the combined DataFrame to a CSV file
# combined_df.to_csv('combined_variants_with_normal.csv', index=False)
# print(f"Exported {len(combined_df)} rows to combined_variants_with_normal.csv")
```

```python
# Sort the combined DataFrame by snp_id to group variants with the same SNP ID together
combined_df = combined_df.sort_values(by='snp_id')

# Reset the index after sorting
combined_df.reset_index(drop=True, inplace=True)

# Export the sorted DataFrame to a CSV file
# combined_df.to_csv('combined_variants_sorted_by_snp_id.csv', index=False)
# print(f"Exported {len(combined_df)} rows to combined_variants_sorted_by_snp_id.csv")

# Display the first few rows to verify sorting
print("\nFirst few rows of the sorted DataFrame:")
print(combined_df.head())
```

```python
from Bio import Entrez, SeqIO
import re
import time
import io


print("--- Step 1: Setup and Load Data ---")

# --- Configuration ---
Entrez.email = "your.email@example.com" # ALWAYS provide your email
Entrez.api_key = "REDACTED"  # Optional, but recommended
entrez_delay = 0.2 # Delay between NCBI calls

# --- Global Variables ---
chromosome_acc = "NC_000001.11"
""" 
genomic_region_start = 227683763 - 1000 # this is for SLC19A3
genomic_region_end = 227718028 + 2000
"""
genomic_region_start = 169463909 - 2000 #slc19a2
genomic_region_end = 169485970 + 2000

print(f"--- Fetching WT Genomic DNA Region ---")
print(f"Fetching {chromosome_acc} from {genomic_region_start} to {genomic_region_end}...")
wt_genomic_sequence_region = None
try:
    time.sleep(entrez_delay)
    handle_dna = Entrez.efetch(db="nucleotide",
                               id=chromosome_acc,
                               rettype="fasta",
                               retmode="text",
                               seq_start=genomic_region_start,
                               seq_stop=genomic_region_end,
                               strand=1)
    dna_record = SeqIO.read(io.StringIO(handle_dna.read()), "fasta")
    handle_dna.close()
    wt_genomic_sequence_region = str(dna_record.seq).upper()
    print(f"Successfully fetched WT genomic region sequence (Length: {len(wt_genomic_sequence_region)}).")
    expected_len = genomic_region_end - genomic_region_start + 1
    if len(wt_genomic_sequence_region) != expected_len:
        print(f"WARNING: Fetched genomic sequence length ({len(wt_genomic_sequence_region)}) does not match expected length ({expected_len}).")
except Exception as e:
    print(f"ERROR fetching WT genomic DNA: {e}")
    print("Cannot proceed without WT genomic sequence.")
    wt_genomic_sequence_region = None

def get_snp_api_text(snp_id_numeric):
    """
    Fetches SNP info text (requesting XML format) from NCBI Efetch.

    Args:
        snp_id_numeric (str): The numeric part of the rsID.

    Returns:
        str: Raw text content from API, or None on failure.
    """
    print(f"      API Call: Fetching text for rs{snp_id_numeric}")
    raw_text_output = None
    if not snp_id_numeric or not snp_id_numeric.isdigit():
         print(f"      API Call ERROR: Invalid numeric SNP ID: '{snp_id_numeric}'")
         return None
    try:
        time.sleep(entrez_delay) # Respect NCBI rate limits
        handle = Entrez.efetch(db="snp", id=snp_id_numeric, rettype="xml", retmode="xml")
        output_bytes = handle.read()
        handle.close()
        if output_bytes:
            raw_text_output = output_bytes.decode('utf-8')
            print(f"      API Call SUCCESS: Fetched text (length {len(raw_text_output)}).")
        else:
             print(f"      API Call WARNING: No data received for rs{snp_id_numeric}.")
    except Entrez.HTTPError as http_err:
         print(f"      API Call NETWORK ERROR for rs{snp_id_numeric}: {http_err}")
    except Exception as e:
         print(f"      API Call GENERAL ERROR for rs{snp_id_numeric}: {e}")
    return raw_text_output
```

```python
def parse_spdi_alleles_from_text(api_text, chromosome_acc):
    """
    Parse SPDI allele notation from NCBI SNP API XML response.
    
    Args:
        api_text (str): Raw XML text from NCBI SNP API
        chromosome_acc (str): Chromosome accession to match (e.g., 'NC_000001.11')
    
    Returns:
        tuple: (reference_allele, alternate_allele) or (None, None) if parsing fails
    """
    if not api_text or not chromosome_acc:
        print("      SPDI Parse ERROR: Missing API text or chromosome accession")
        return None, None
    
    try:
        # Look for SPDI pattern in the XML text
        # Format: sequence_id:position:deleted_sequence:inserted_sequence
        import re
        
        # Pattern to match SPDI notation with the specific chromosome accession
        pattern = rf'{chromosome_acc}:(\d+):([ACGT]*):([ACGT]*)'
        
        # Find all matches
        matches = re.findall(pattern, api_text)
        
        if not matches:
            print(f"      SPDI Parse WARNING: No SPDI notation found for {chromosome_acc}")
            return None, None
        
        # Use the first match (could be refined to select the canonical one if needed)
        position, ref_allele, alt_allele = matches[0]
        
        # Validate alleles
        if ref_allele == '' and alt_allele == '':
            print("      SPDI Parse ERROR: Both reference and alternate alleles are empty")
            return None, None
            
        print(f"      SPDI Parse SUCCESS: Found ref='{ref_allele}', alt='{alt_allele}'")
        return ref_allele, alt_allele
        
    except Exception as e:
        print(f"      SPDI Parse ERROR: {e}")
        return None, None

# --- [PREVIOUS CODE: Imports, Setup, WT Fetch, Helper Functions] ---
# --- [PLACEHOLDER: Ensure combined_df is loaded] ---
# --- [Check if required variables exist] ---

# --- Validation and Correction Loop ---
corrections_made = 0
validation_failures = 0 # Counter for persistent/uncorrected errors in non-skipped rows
skipped_insertions = 0 # Counter for rows skipped due to original == '-'

# Ensure required columns are present in combined_df
required_cols = ['pos', 'original', 'mutant', 'snp_id', 'clinical_significance']
if not all(col in combined_df.columns for col in required_cols):
    print(f"ERROR: DataFrame 'combined_df' missing one or more required columns: {required_cols}")
    # Optionally exit or handle error
else:
    # Filter rows where clinical_significance is not 'normal'
    target_mask = ~combined_df['clinical_significance'].astype(str).str.lower().eq('normal')
    print(f"Processing {target_mask.sum()} rows where clinical_significance is not 'normal'.")

    # Use indices for direct update with .loc on combined_df
    for index in combined_df[target_mask].index:
        row = combined_df.loc[index] # Get row data

        try:
            # Extract original allele first to check for '-'
            original_allele = str(row['original']).upper() if pd.notna(row['original']) else ''
            # Still extract pos and snp_id for potential logging later, even if skipping
            pos = int(row['pos'])
            snp_id = str(row['snp_id']) if pd.notna(row['snp_id']) else ''

        except (ValueError, TypeError) as e:
             # If basic info can't be read, skip and maybe count as failure
             print(f"  SKIP (Index {index}): Error converting basic row data (pos/original/snp_id): {e}.")
             # validation_failures += 1 # Decide if this counts as failure
             continue # Skip this row

        # --- *** ADDED: Skip rows where original allele is '-' *** ---
        if original_allele == '-':
            skipped_insertions += 1
            # Optional: print a message if you want to see which rows are skipped
            # print(f"  SKIP (snp {snp_id}, pos {pos}): Row skipped because original allele is '-'.")
            continue # Immediately go to the next iteration of the loop
        # --- *** END OF ADDED SKIP LOGIC *** ---

        # --- If we are here, original_allele is NOT '-' ---
        # Now extract mutant allele if needed for later steps
        try:
            mutant_allele = str(row['mutant']).upper() if pd.notna(row['mutant']) else ''
        except (ValueError, TypeError) as e:
            print(f"  SKIP (snp {snp_id}, pos {pos}): Error converting mutant allele: {e}.")
            validation_failures += 1 # Count failure if mutant allele bad for non '-' row
            continue

        # --- Basic Calculation & Bounds Check (only for non '-' rows) ---
        index_0based = pos - genomic_region_start
        ref_len = len(original_allele) # Will not be 0 here unless original was empty string ''

        # Handle potentially empty original allele (if not NaN and not '-')
        if ref_len == 0:
            # print(f"  SKIP (snp {snp_id}, pos {pos}): Empty 'original' allele (and not '-').")
            continue

        # Bounds checks
        if not (0 <= index_0based < len(wt_genomic_sequence_region)):
            print(f"  FAIL (snp {snp_id}, pos {pos}): Position {pos} maps to index {index_0based}, outside fetched WT bounds.")
            validation_failures += 1
            continue
        if index_0based + ref_len > len(wt_genomic_sequence_region):
             print(f"  FAIL (snp {snp_id}, pos {pos}): Original allele '{original_allele}' (len {ref_len}) at index {index_0based} extends beyond fetched WT length.")
             validation_failures += 1
             continue

        # --- Compare (only for non '-' rows) ---
        ref_from_genome = wt_genomic_sequence_region[index_0based : index_0based + ref_len].upper()

        if ref_from_genome != original_allele:
            # --- Mismatch: Attempt Strategy #2 (API Fix) ---
            mismatch_detail = f"Mismatch at pos {pos}: DataFrame 'original'='{original_allele}', Genome='{ref_from_genome}'."
            print(f"  WARN (snp {snp_id}): {mismatch_detail} Trying API Fix...")

            numeric_snp_id = None
            # Extract numeric SNP ID
            if snp_id and isinstance(snp_id, str):
                 if snp_id.lower().startswith('rs') and snp_id[2:].isdigit():
                     numeric_snp_id = snp_id[2:]
                 elif snp_id.isdigit():
                     numeric_snp_id = snp_id

            if not numeric_snp_id:
                print(f"  FAIL (snp {snp_id}): Invalid or missing snp_id ('{snp_id}') for API lookup. Fix attempt failed.")
                validation_failures += 1
                continue

            api_raw_text = get_snp_api_text(numeric_snp_id)

            if api_raw_text:
                spdi_ref, spdi_alt = parse_spdi_alleles_from_text(api_raw_text, chromosome_acc)

                if spdi_ref is not None: # SPDI parsed
                    spdi_ref_len = len(spdi_ref)

                    if spdi_ref_len == 0:
                         print(f"  FAIL (snp {snp_id}): API Fix failed. SPDI Ref allele from API is empty.")
                         validation_failures += 1; continue

                    # Re-check bounds using SPDI ref length
                    if index_0based + spdi_ref_len > len(wt_genomic_sequence_region):
                         print(f"  FAIL (snp {snp_id}): API Fix failed. SPDI Ref '{spdi_ref}' extends beyond fetched WT.")
                         validation_failures += 1; continue

                    # Re-validate WT genome segment vs SPDI Ref
                    expected_ref_spdi = wt_genomic_sequence_region[index_0based : index_0based + spdi_ref_len].upper()

                    if expected_ref_spdi == spdi_ref.upper():
                        print(f"  SUCCESS (snp {snp_id}): API Fix corrected alleles based on SPDI. Updating DataFrame: REF='{spdi_ref}', ALT='{spdi_alt}'.")
                        # Update DataFrame
                        combined_df.loc[index, 'original'] = spdi_ref
                        combined_df.loc[index, 'mutant'] = spdi_alt
                        corrections_made += 1
                    else:
                        print(f"  FAIL (snp {snp_id}): API Fix failed re-check. WT base(s) '{expected_ref_spdi}' != SPDI Ref '{spdi_ref}'. Alleles not corrected.")
                        validation_failures += 1; continue
                else: # SPDI parse failed
                    print(f"  FAIL (snp {snp_id}): API Fix failed (SPDI pattern not found in API response).")
                    validation_failures += 1; continue
            else: # API fetch failed
                print(f"  FAIL (snp {snp_id}): API Fix failed (Error fetching data from API).")
                validation_failures += 1; continue
        # else: Initial match for non '-' row, do nothing

    print(f"\n--- Validation Finished ---")
    print(f"Rows skipped because 'original' allele was '-': {skipped_insertions}")
    print(f"Total Allele Corrections Made via API SPDI lookup (for non '-' rows): {corrections_made}")
    print(f"Total Persistent/Uncorrected Errors (Mismatches or Failures for non '-' rows): {validation_failures}")
    print("\nDataFrame head after validation/correction attempt:")
    # Display relevant columns
    print(combined_df[['#chr', 'pos', 'snp_id', 'original', 'mutant', 'clinical_significance']].head(10))

# --- End of Validation Step ---
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

columns = ['MGP_freq', 'MGP_is_mentioned', 'HGDP_Stanford_freq', 'HGDP_Stanford_is_mentioned',
           'FINRISK_freq', 'FINRISK_is_mentioned', 'Chileans_freq', 'Chileans_is_mentioned',
           'GENOME_DK_freq', 'GENOME_DK_is_mentioned', 'PharmGKB_freq', 'PharmGKB_is_mentioned',
           'GnomAD_exomes_freq', 'GnomAD_exomes_is_mentioned', 'SGDP_PRJ_freq', 'SGDP_PRJ_is_mentioned',
           'ExAC_freq', 'ExAC_is_mentioned', 'PAGE_STUDY_freq', 'PAGE_STUDY_is_mentioned',
           'Korea1K_freq', 'Korea1K_is_mentioned', 'Estonian_freq', 'Estonian_is_mentioned',
           'ALSPAC_freq', 'ALSPAC_is_mentioned', 'Korea4K_freq', 'Korea4K_is_mentioned',
           'KOREAN_freq', 'KOREAN_is_mentioned', 'GENOGRAPHIC_freq', 'GENOGRAPHIC_is_mentioned',
           'Daghestan_freq', 'Daghestan_is_mentioned', 'GoNL_freq', 'GoNL_is_mentioned',
           'TOMMO_freq', 'TOMMO_is_mentioned', '1000Genomes_30X_freq', '1000Genomes_30X_is_mentioned',
           'HapMap_freq', 'HapMap_is_mentioned', '1000Genomes_freq', '1000Genomes_is_mentioned',
           'TWINSUK_freq', 'TWINSUK_is_mentioned', 'NorthernSweden_freq', 'NorthernSweden_is_mentioned',
           'PRJEB36033_freq', 'PRJEB36033_is_mentioned', 'GoESP_freq', 'GoESP_is_mentioned',
           'Vietnamese_freq', 'Vietnamese_is_mentioned', 'TOPMED_freq', 'TOPMED_is_mentioned',
           'PRJEB37584_freq', 'PRJEB37584_is_mentioned', 'ALFA_freq', 'ALFA_is_mentioned',
           'GnomAD_genomes_freq', 'GnomAD_genomes_is_mentioned', 'Siberian_freq', 'Siberian_is_mentioned',
           'Qatari_freq', 'Qatari_is_mentioned']

columns_ones_only = ['by-frequency', 'by-alfa', 'by-cluster']


import numpy as np
import pandas as pd

# Use a random number generator for reproducibility
rng = np.random.default_rng(42)

# --- Impute *_freq values for rows labeled 'normal' using ONLY 'benign' rows as reference ---

p_force_zero = 0.50  # probability to force an imputed freq to zero

mask_normal = combined_df['clinical_significance'].eq('normal')
mask_benign = combined_df['clinical_significance'].eq('benign')
freq_cols = [c for c in combined_df.columns if c.endswith('_freq')]

# Reference stats from benign rows ONLY
ref_source = combined_df.loc[mask_benign, freq_cols].apply(pd.to_numeric, errors='coerce')

if ref_source.shape[0] == 0:
    # Fallback: if no benign rows, optionally use previous strategy (all non-normal) or skip.
    # Here we choose to skip with a warning-like print.
    print("Warning: No benign rows found; skipping imputation.")
    perform_impute = False
else:
    means = ref_source.mean()
    stds = ref_source.std().fillna(0)
    perform_impute = True

n_to_impute = mask_normal.sum()

if perform_impute and n_to_impute > 0:
    # Vectorized draw for efficiency
    # Build arrays of means/stds aligned to freq_cols
    mean_vec = means.reindex(freq_cols).to_numpy()
    std_vec = stds.reindex(freq_cols).to_numpy()

    # For columns with NaN mean (all NaN in benign), we won't change those values
    valid_mask = ~np.isnan(mean_vec)

    # Prepare an (n_to_impute, n_freq_cols) matrix initialized with existing values
    impute_block = combined_df.loc[mask_normal, freq_cols].apply(pd.to_numeric, errors='coerce').to_numpy(copy=True)

    if valid_mask.any():
        # Draw normals where std > 0 else fill mean
        # Expand mean/std to row count
        means_expanded = np.tile(mean_vec, (n_to_impute, 1))
        stds_expanded = np.tile(std_vec, (n_to_impute, 1))

        # Draw standard normals then scale; where std==0 just use mean
        rand_normals = rng.standard_normal(size=(n_to_impute, len(freq_cols)))
        new_vals = means_expanded + rand_normals * stds_expanded
        new_vals[:, std_vec == 0] = means_expanded[:, std_vec == 0]

        # Keep only valid columns
        new_vals[:, ~valid_mask] = impute_block[:, ~valid_mask]  # leave untouched

        # Clip to [0,1]
        new_vals = np.clip(new_vals, 0.0, 1.0)

        # Force zeros independently per cell
        zero_mask = rng.random(size=new_vals.shape) < p_force_zero
        new_vals[zero_mask] = 0.0

        # Assign back
        combined_df.loc[mask_normal, freq_cols] = new_vals

# --- Recalculate derived columns ---

freq_mat = combined_df[freq_cols].apply(pd.to_numeric, errors='coerce')

# *_is_mentioned (1 if freq > 0)
mention_cols = []
for col in freq_cols:
    mention_col = col.replace('_freq', '_is_mentioned')
    if mention_col in combined_df.columns:
        mention_cols.append(mention_col)
        combined_df[mention_col] = (freq_mat[col] > 0).astype(int)

combined_df['Total_Studies'] = combined_df[mention_cols].sum(axis=1)

# Mask zeros (treat as absent) for summary stats
masked_freq = freq_mat.where(freq_mat > 0)

combined_df['Avg_Frequency'] = masked_freq.mean(axis=1)
combined_df['Max_Frequency'] = masked_freq.max(axis=1)
combined_df['Min_Frequency'] = masked_freq.min(axis=1)

summary_cols = ['Avg_Frequency', 'Max_Frequency', 'Min_Frequency']
combined_df.loc[combined_df['Total_Studies'] == 0, summary_cols] = 0.0

cols_set_one = ['by-frequency', 'by-alfa', 'by-cluster']

combined_df.loc[mask_normal, cols_set_one] = 1
```

```python
import numpy as np
def extract_studies_and_create_columns(df):
    unique_study_names = set()
    for freq_string in df['frequency'].dropna():
        study_reports = freq_string.split('|')
        for report in study_reports:
            parts = report.split(':')
            if len(parts) == 4:
                study_name = parts[3]
                unique_study_names.add(study_name)
    
    # Removed sorting as per your instruction
    unique_study_names_list = list(unique_study_names) 
    
    for study in unique_study_names_list: # Use the unsorted list
        df[f'{study}_freq'] = 0.0
        df[f'{study}_is_mentioned'] = 0

    df['Avg_Frequency'] = 0.0
    df['Max_Frequency'] = 0.0
    df['Min_Frequency'] = 0.0
    df['Total_Studies'] = 0

    return df, unique_study_names_list # Return the unsorted list

def populate_frequencies_and_summaries(df, unique_study_names):
    df_processed = df.copy()

    for index, row in df_processed.iterrows():
        mutant_allele = row['mutant']
        clinical_significance = row['clinical_significance']
        
        if pd.isna(clinical_significance) or clinical_significance == '':
            clinical_significance = ''

        if clinical_significance != 'normal':
            current_freqs_for_summary = []
            if pd.notna(row['frequency']) and str(row['frequency']).strip():
                freq_entries = str(row['frequency']).split('|')
                for entry in freq_entries:
                    parts = entry.split(':')
                    if len(parts) == 4:
                        allele_in_freq_string = parts[0]
                        freq_str = parts[1]
                        study_name_from_freq = parts[3]

                        if study_name_from_freq in unique_study_names:
                            if allele_in_freq_string == mutant_allele:
                                # Removed try-except for float conversion as per your instruction
                                freq_val = float(freq_str) 
                                df_processed.at[index, f'{study_name_from_freq}_freq'] = freq_val
                                df_processed.at[index, f'{study_name_from_freq}_is_mentioned'] = 1
                                current_freqs_for_summary.append(freq_val)
                                
            
            if current_freqs_for_summary:
                df_processed.at[index, 'Avg_Frequency'] = np.mean(current_freqs_for_summary)
                df_processed.at[index, 'Max_Frequency'] = np.max(current_freqs_for_summary)
                df_processed.at[index, 'Min_Frequency'] = np.min(current_freqs_for_summary)
                df_processed.at[index, 'Total_Studies'] = len(current_freqs_for_summary)
        
        else:
            for study in unique_study_names:
                df_processed.at[index, f'{study}_freq'] = 1.0
                df_processed.at[index, f'{study}_is_mentioned'] = 1
            
            df_processed.at[index, 'Avg_Frequency'] = 1.0
            df_processed.at[index, 'Max_Frequency'] = 1.0
            df_processed.at[index, 'Min_Frequency'] = 1.0
            df_processed.at[index, 'Total_Studies'] = len(unique_study_names)
            
    return df_processed

combined_df, unique_studies = extract_studies_and_create_columns(combined_df)
combined_df = populate_frequencies_and_summaries(combined_df, unique_studies)
combined_df.head()
```

```python

print("\n--- Step 3: Generating DNA Sequences for Each Row ---")

# --- Ensure prerequisites exist ---
if 'combined_df' not in locals() or combined_df is None:
     print("ERROR: DataFrame 'combined_df' is not available.")
# (Add checks for wt_genomic_sequence_region, genomic_region_start if needed)
# ...
elif 'wt_genomic_sequence_region' not in locals() or wt_genomic_sequence_region is None:
     print("ERROR: WT Genomic sequence ('wt_genomic_sequence_region') is not available.")
elif 'genomic_region_start' not in locals() or genomic_region_start is None:
     print("ERROR: Genomic region start position ('genomic_region_start') is not defined.")
else:
    # --- Initialize ---
    generated_sequences = []
    generation_errors = 0
    wt_len = len(wt_genomic_sequence_region) # Cache length

    # --- Iterate through DataFrame ---
    # Using itertuples provides row index via row.Index
    for row in combined_df.itertuples(index=True):
        generated_seq_for_row = None # Default to None
        row_index = row.Index        # Get the original DataFrame index for error reporting
        snp_id_for_error = 'N/A'     # Default SNP ID if reading fails

        try:
            # --- Read identifiers and critical data FIRST ---
            # Attempt to read SNP ID early for better error messages
            snp_id_for_error = str(getattr(row, 'snp_id', f'Index_{row_index}')).strip()

            # --- Robustly handle 'pos' column ---
            pos_val = getattr(row, 'pos', None) # Get raw value
            if pd.isna(pos_val):
                # Handle NaN values for position
                print(f"  ERROR (snp {snp_id_for_error}, Index {row_index}): 'pos' column is NaN. Skipping row.")
                generation_errors += 1
                generated_sequences.append(None) # Append None as sequence cannot be generated
                continue # Skip to the next row
            try:
                # Attempt integer conversion only if not NaN
                pos = int(pos_val)
            except (ValueError, TypeError):
                # Handle values that are not NaN but cannot be converted to int
                print(f"  ERROR (snp {snp_id_for_error}, Index {row_index}): 'pos' value '{pos_val}' is not a valid integer. Skipping row.")
                generation_errors += 1
                generated_sequences.append(None) # Append None
                continue # Skip to the next row
            # --- End of robust 'pos' handling ---

            # Read other necessary fields
            original_allele = str(getattr(row, 'original', '')).upper()
            mutant_allele = str(getattr(row, 'mutant', '')).upper()
            significance = str(getattr(row, 'clinical_significance', '')).lower()

            # --- Check Significance ---
            if significance == 'normal':
                generated_seq_for_row = wt_genomic_sequence_region
            else:
                # --- Process non-normal variants ---
                index_0based = pos - genomic_region_start

                # --- Handle Insertion (original == '-') ---
                if original_allele == '-':
                    if not (0 <= index_0based <= wt_len):
                         print(f"  ERROR (snp {snp_id_for_error}, pos {pos}, Index {row_index}): Insertion index {index_0based} out of bounds [0, {wt_len}].")
                         generation_errors += 1
                    else:
                         generated_seq_for_row = wt_genomic_sequence_region[:index_0based] + \
                                                 mutant_allele + \
                                                 wt_genomic_sequence_region[index_0based:]

                # --- Handle Substitution/Deletion (original != '-' and not empty) ---
                elif original_allele:
                    ref_len = len(original_allele)
                    if not (0 <= index_0based < wt_len and index_0based + ref_len <= wt_len):
                        print(f"  ERROR (snp {snp_id_for_error}, pos {pos}, Index {row_index}): Original allele '{original_allele}' segment out of bounds for WT sequence.")
                        generation_errors += 1
                    else:
                        ref_from_genome = wt_genomic_sequence_region[index_0based : index_0based + ref_len].upper()
                        if ref_from_genome == original_allele:
                            # Match: Apply mutation
                            if mutant_allele == '-': # Deletion
                                generated_seq_for_row = wt_genomic_sequence_region[:index_0based] + \
                                                        wt_genomic_sequence_region[index_0based + ref_len:]
                            else: # Substitution
                                generated_seq_for_row = wt_genomic_sequence_region[:index_0based] + \
                                                        mutant_allele + \
                                                        wt_genomic_sequence_region[index_0based + ref_len:]
                        else:
                            # Mismatch: This is the validation error
                            print(f"  ERROR (snp {snp_id_for_error}, pos {pos}, Index {row_index}): Validation failed. Expected WT '{original_allele}' but found '{ref_from_genome}'.")
                            generation_errors += 1
                            # generated_seq_for_row remains None
                else:
                    # Handle case where original is empty but not '-'
                    print(f"  WARN (snp {snp_id_for_error}, pos {pos}, Index {row_index}): Original allele is empty but not '-'. Cannot process.")
                    generation_errors += 1
                    # generated_seq_for_row remains None

        except Exception as e:
            # Catch any other unexpected errors (should be less frequent now)
            print(f"  CRITICAL ERROR processing row (SNP/ID: {snp_id_for_error}, Index: {row_index}): {e}")
            generation_errors += 1
            generated_seq_for_row = None # Ensure None on critical error

        # Append result for this row
        generated_sequences.append(generated_seq_for_row)

    # --- Add the new column to the DataFrame ---
    if len(generated_sequences) == len(combined_df):
        combined_df['generated_dna'] = generated_sequences
        print(f"\n'generated_dna' column added successfully.")
    else:
        print(f"\nERROR: Length mismatch! Generated sequences ({len(generated_sequences)}) != DataFrame rows ({len(combined_df)}). Column not added.")
        generation_errors = len(combined_df) # Indicate major failure

    # --- Report Errors ---
    print(f"\n--- Sequence Generation Finished ---")
    print(f"Total errors encountered during sequence generation: {generation_errors}")

    # --- Display Head ---
    if 'generated_dna' in combined_df.columns:
        print("\nDataFrame head including 'generated_dna' column info:")
        display_df = combined_df[['#chr', 'pos', 'snp_id', 'original', 'mutant', 'clinical_significance']].copy()
        display_df['generated_dna_type'] = combined_df['generated_dna'].apply(lambda x: type(x).__name__)
        display_df['generated_dna_len'] = combined_df['generated_dna'].apply(lambda x: len(x) if isinstance(x, str) else 0)
        print(display_df.head(10))
        failed_rows = combined_df['generated_dna'].isna().sum()
        if failed_rows > 0:
             print(f"\nNote: {failed_rows} row(s) have 'None'/'NaN' in 'generated_dna' due to errors.")
    else:
        print("\n'generated_dna' column was not added due to errors.")
```

```python
mRNA_Ready = combined_df.copy()
# Create a new column for generated mRNA sequences
mRNA_Ready['generated_mRNA'] = None

def calculate_relative_position(pos, exon_start):
    return int(pos) - int(exon_start)
     
# Function to check if a position is within any exon
def is_position_in_exon(pos):
    '''
    Check if a position is within any exon and return the exon number, genomic start, genomic end, sequence, and relative position.
    
    Args:
        pos (int): Genomic position to check
        
    Returns:
        tuple: (True/False, exon_index (1-based), genomic_start, genomic_end, sequence, relative_position)
    '''
    for index, row in exons_df.iterrows():
        if row['Genomic_Start'] <= pos <= row['Genomic_End']:
            relative_position = calculate_relative_position(pos, row['Genomic_Start'])
            return True, index+1, row['Genomic_Start'], row['Genomic_End'], row['Sequence'], relative_position
    return False, None, None, None, None, None


def fuse_exons(exon_index_1_based, sequence):
    # fuse the exons to create the full sequence
    exons_fused = ''
    for index,row in exons_df.iterrows():
        if index+1 != exon_index_1_based:
            exons_fused += row['Sequence']
        else:
            exons_fused += sequence
    return exons_fused

# Initialize counters outside the function
substitutions = 0
delted_stuff = 0
inserted_in_exon = 0

def mutate_exon(exon_sequence, relative_position, original, mutant):
    global substitutions, delted_stuff, inserted_in_exon
    mutated_exon = list(exon_sequence)
    
    if mutant != '-' and original != '-':
        # Handle substitutions of different lengths
        # First remove the original sequence
        del mutated_exon[relative_position:relative_position + len(original)]
        # Then insert the new sequence
        for i, char in enumerate(mutant):
            mutated_exon.insert(relative_position + i, char)
        substitutions += 1

    elif mutant == '-':  # Deletion
        # Remove the characters at the position
        del mutated_exon[relative_position:relative_position + len(original)]
        delted_stuff += 1
    
    elif original == '-':  # Insertion
        # Just insert the new sequence at the position
        for i, char in enumerate(mutant):
            mutated_exon.insert(relative_position + i, char)
        inserted_in_exon += 1
    
    return ''.join(mutated_exon)

def mutate_reverse_complemnet_and_return_mRNA(exon_index_1_based, exon_sequence, relative_position, original, mutant):
    # this takes the exon sequence on positive strand, mutate it and joins with other exons, then reverse complements it converts it to mRNA
    mutated_exon = mutate_exon(exon_sequence, relative_position, original, mutant)
    mutated_axons_on_template_strand = fuse_exons(exon_index_1_based, mutated_exon)
    tranlsated_sequence = str(Seq(mutated_axons_on_template_strand).reverse_complement())
    mRNA = tranlsated_sequence.replace('T', 'U')
    return mRNA

# Iterate through each row in the dataframe and generate the mRNA sequence for each row (complement based)
Non_matches = 0
skipped_cause_of_normal = 0
skipped_cause_of_not_in_exon = 0
matched_in_exon_and_mutated = 0 

for idx, row in mRNA_Ready.iterrows():
    if row['clinical_significance'] == 'normal':
        # For normal variants, use the wild type mRNA
        mRNA_Ready.at[idx, 'generated_mRNA'] = mRNA_WT
        skipped_cause_of_normal += 1
    else:
        # For non-normal variants, check if position is within any exon
        pos = int(row['pos'])
        is_in_exon, exon_index_1_based, genomic_start, genomic_end, exon_sequence, relative_position = is_position_in_exon(pos)
        if not is_in_exon:
            mRNA_Ready.at[idx, 'generated_mRNA'] = mRNA_WT
            skipped_cause_of_not_in_exon += 1
        else:
            len_of_original = len(row['original'])
            if row['original'] == '-':
                mRNA = mutate_reverse_complemnet_and_return_mRNA(exon_index_1_based, exon_sequence, relative_position, row['original'], row['mutant'])
                mRNA_Ready.at[idx, 'generated_mRNA'] = mRNA
            
            elif exon_sequence[relative_position:relative_position+len_of_original] == row['original']:
                print(f'''Mutation of SNP_ID: {row["snp_id"]} exists in exon {exon_index_1_based}
                      And it Matches, we are going to change {row['original']} to {row['mutant']}''')
                
                mRNA = mutate_reverse_complemnet_and_return_mRNA(exon_index_1_based, exon_sequence, relative_position, row['original'], row['mutant'])
                mRNA_Ready.at[idx, 'generated_mRNA'] = mRNA
                matched_in_exon_and_mutated += 1
            else:
                print(f'''Mutation of SNP_ID: {row["snp_id"]} exists in exon {exon_index_1_based}
                      And it does not match, expected {row['original']} and found {exon_sequence[relative_position:relative_position+len_of_original]}''')
                Non_matches += 1
                mRNA_Ready.at[idx, 'generated_mRNA'] = mRNA_WT  # Use wild type for non-matches
                
total = matched_in_exon_and_mutated + skipped_cause_of_normal + skipped_cause_of_not_in_exon + Non_matches
# Print summary of mRNA generation
print(f'''generated mRNA on negative strand, matched in exon and mutated: {matched_in_exon_and_mutated}, 
      got normal mRNA: {skipped_cause_of_normal},
      skipped because not in exon: {skipped_cause_of_not_in_exon},
      non matches: {Non_matches},
      substitutions: {substitutions},
      inserted in exon: {inserted_in_exon},
      deleted in exon: {delted_stuff},
      total: {total}
''')
```

```python
# Dependencies
from Bio import Entrez, SeqIO
from Bio.Seq import Seq
import pandas as pd
import math
from collections import Counter

# --- 1) Get CDS bounds from existing vars or GenBank (NM_006996.3) ---
def get_cds_from_globals_or_genbank(accession="NM_006996.3", email="your_email@example.com", api_key=None):
    cds_start = globals().get("cds_start_index", None)
    cds_end = globals().get("cds_end_index", None)
    ref_mrna = globals().get("reference_mrna", None) or globals().get("mRNA_reference", None) or globals().get("ref_mrna", None)

    if cds_start is not None and cds_end is not None and isinstance(ref_mrna, str) and len(ref_mrna) > 0:
        return int(cds_start), int(globals()["cds_end_index"]), ref_mrna, "globals"

    # Fallback: fetch NM_006996.3 GenBank
    Entrez.email = email
    if api_key:
        Entrez.api_key = api_key
    with Entrez.efetch(db="nucleotide", id=accession, rettype="gb", retmode="text") as handle:
        record = SeqIO.read(handle, "genbank")

    cds_feats = [f for f in record.features if f.type == "CDS"]
    assert len(cds_feats) >= 1, "No CDS feature found in NM_006996.3"
    cds = cds_feats[0]
    cds_start = int(cds.location.start)
    cds_end = int(cds.location.end)
    ref_mrna = str(record.seq).upper()  # NCBI NM_... is DNA alphabet (T)

    return cds_start, cds_end, ref_mrna, "genbank"

cds_start_idx, cds_end_idx, reference_mrna_DNA, source = get_cds_from_globals_or_genbank()

# Confirm AUG on reference at CDS start
ref_rna = reference_mrna_DNA.replace("T", "U").upper()
ref_start_codon = ref_rna[cds_start_idx:cds_start_idx+3] if cds_start_idx+3 <= len(ref_rna) else ""
print(f"[Reference {source}] CDS start index: {cds_start_idx}, end: {cds_end_idx}, length: {len(ref_rna)}")
print("Reference CDS-start codon (RNA letters):", ref_start_codon, 
      "context:", ref_rna[max(0, cds_start_idx-9):min(len(ref_rna), cds_start_idx+12)])
assert ref_start_codon == "AUG", "Reference CDS start is not AUG on the reference mRNA. Check transcript alignment."

# --- 2) Auto-detect your DataFrame and mRNA column ---
def pick_mrna_df_and_col():
    # Prefer common names if present
    preferred_df_names = ["mRNA_Ready", "mrna_df", "df", "final_df", "variants_df", "data", "dataset"]
    for name in preferred_df_names:
        obj = globals().get(name, None)
        if isinstance(obj, pd.DataFrame):
            cols = [c.lower() for c in obj.columns]
            if any("mrna" in c for c in cols):
                # pick the most likely column
                for candidate in ["generated_mRNA", "generated_mrna", "mRNA", "mrna"]:
                    for col in obj.columns:
                        if col.lower() == candidate.lower():
                            return obj, col
                # fallback: first column containing 'mrna'
                for col in obj.columns:
                    if "mrna" in col.lower():
                        return obj, col

    # Broad scan of all DataFrames in globals
    for name, obj in list(globals().items()):
        if isinstance(obj, pd.DataFrame):
            for col in obj.columns:
                if "mrna" in col.lower():
                    return obj, col
    raise RuntimeError("Could not find a DataFrame with an mRNA column. Please set df_mrna and col_mrna manually.")

df_mrna, col_mrna = pick_mrna_df_and_col()
print(f"Using DataFrame with shape {df_mrna.shape}, column '{col_mrna}' for mRNA sequences.")

# --- 3) Scan sequences for AUG at CDS start (and nearby in-frame) ---
radius_nt = 30  # search window (+/-) in nucleotides, in-frame only
total = len(df_mrna)
exact_at_start = 0
inframe_nearby = 0
aug_elsewhere = 0
no_aug_anywhere = 0
too_short = 0

examples = {
    "not_exact_but_inframe_near": [],
    "no_inframe_near_but_elsewhere": [],
    "no_aug_anywhere": [],
    "too_short": []
}

def inframe_positions(center, radius, seq_len):
    start = max(0, center - radius)
    end = min(seq_len - 3, center + radius)
    # only in-frame positions relative to cds_start
    for i in range(start, end + 1):
        if (i - center) % 3 == 0:
            yield i

for idx, row in df_mrna.iterrows():
    s = row[col_mrna]
    if not isinstance(s, str) or len(s) < cds_start_idx + 3:
        too_short += 1
        if len(examples["too_short"]) < 3:
            examples["too_short"].append({"idx": idx, "len": 0 if not isinstance(s, str) else len(s)})
        continue
    rna = s.upper().replace("T", "U")

    # Exact check
    codon = rna[cds_start_idx:cds_start_idx+3]
    if codon == "AUG":
        exact_at_start += 1
        continue

    # In-frame nearby search
    nearest = None
    best_dist = math.inf
    for i in inframe_positions(cds_start_idx, radius_nt, len(rna)):
        if rna[i:i+3] == "AUG":
            d = abs(i - cds_start_idx)
            if d < best_dist:
                best_dist = d
                nearest = i
    if nearest is not None:
        inframe_nearby += 1
        if len(examples["not_exact_but_inframe_near"]) < 3:
            context = rna[max(0, nearest-9):min(len(rna), nearest+12)]
            examples["not_exact_but_inframe_near"].append({
                "idx": idx, "nearest_pos": nearest, "dist_nt": best_dist, "context": context
            })
        continue

    # Elsewhere in sequence?
    if "AUG" in rna:
        aug_elsewhere += 1
        if len(examples["no_inframe_near_but_elsewhere"]) < 3:
            # show first AUG elsewhere
            pos = rna.find("AUG")
            context = rna[max(0, pos-9):min(len(rna), pos+12)]
            examples["no_inframe_near_but_elsewhere"].append({
                "idx": idx, "first_aug_pos": pos, "context": context
            })
    else:
        no_aug_anywhere += 1
        if len(examples["no_aug_anywhere"]) < 3:
            examples["no_aug_anywhere"].append({"idx": idx})

# --- 4) Report ---
print("\nAUG alignment summary (relative to reference CDS start):")
print(f"Total sequences: {total}")
print(f"- Exact AUG at CDS start: {exact_at_start} ({exact_at_start/total:.1%})")
print(f"- In-frame AUG within ±{radius_nt} nt (not exact): {inframe_nearby} ({inframe_nearby/total:.1%})")
print(f"- AUG present but not in-frame near start: {aug_elsewhere} ({aug_elsewhere/total:.1%})")
print(f"- No AUG anywhere in sequence: {no_aug_anywhere} ({no_aug_anywhere/total:.1%})")
print(f"- Too short (< CDS start + 3): {too_short} ({too_short/total:.1%})")

def print_examples(title, lst):
    if lst:
        print(f"\nExamples — {title}:")
        for ex in lst:
            print(ex)

print_examples("not exact but in-frame nearby", examples["not_exact_but_inframe_near"])
print_examples("no in-frame near but has AUG elsewhere", examples["no_inframe_near_but_elsewhere"])
print_examples("no AUG anywhere", examples["no_aug_anywhere"])
print_examples("too short", examples["too_short"])
```

```python
import re
import os
import time
import pandas as pd
from Bio.Seq import Seq
from Bio import Entrez, SeqIO
from warnings import filterwarnings

# --- 1. Configuration & Setup ---
# **CUSTOMIZE** your email for NCBI Entrez
Entrez.email = "your.email@example.com"
entrez_delay = 0.5 # Delay between NCBI calls to be polite

aug_search_activations_count = 0
# early_stop_count will be calculated after translation

# --- 2. Variables assumed loaded (Ensure mrna_acc and mrna_df are defined) ---
# Example definitions (replace with your actual variables)
mrna_acc = "NM_006996.3" #SLC19A2
# Make sure mRNA_Ready is loaded from your data source, e.g., a CSV
# mrna_df = mRNA_Ready.copy()
# Using a placeholder if mRNA_Ready is not defined:
try:
    mrna_df = mRNA_Ready.copy()
except NameError:
    print("Warning: `mRNA_Ready` not found. Please ensure it is loaded.")
    mrna_df = None # Set to None to prevent errors later

# --- 3. Fetch Reference CDS Info ---
print(f"\n--- Fetching GenBank record for {mrna_acc} to find reference CDS info ---")
cds_start = None
cds_end = None
wt_protein_sequence_ref = None
wt_rna_sequence_ref = None
wt_kozak_sequence = None
wt_start_in_window = None
WT_KOZAK_WINDOW_NT = 9 # How many bases to look at on each side of AUG's 'A'

if mrna_acc:
    try:
        time.sleep(entrez_delay)
        handle_gb = Entrez.efetch(db="nucleotide", id=mrna_acc, rettype="gb", retmode="text")
        record = SeqIO.read(handle_gb, "genbank")
        handle_gb.close()
        print(f"Fetched GenBank record for {mrna_acc}.")

        wt_rna_sequence_ref = str(record.seq).upper().replace('T', 'U') # Get WT sequence as RNA

        for feature in record.features:
            if feature.type == "CDS":
                cds_start = int(feature.location.start)
                cds_end = int(feature.location.end)
                print(f"  Found reference CDS feature: Start Index={cds_start}, End Index={cds_end}")

                if "translation" in feature.qualifiers:
                    wt_protein_sequence_ref = feature.qualifiers['translation'][0]

                # Define the reference Kozak context and its true offset
                wt_kozak_sequence = wt_rna_sequence_ref[max(0, cds_start - WT_KOZAK_WINDOW_NT) : cds_start + 3 + WT_KOZAK_WINDOW_NT]
                wt_start_in_window = min(cds_start, WT_KOZAK_WINDOW_NT)
                print(f"  Reference Kozak context: '{wt_kozak_sequence}' (AUG anchor at index {wt_start_in_window})")
                break # Stop after finding the first CDS

        if cds_start is None: print(f"  WARNING: No CDS feature found for {mrna_acc}.")

    except Exception as e:
        print(f"  ERROR fetching or parsing GenBank record for CDS info: {e}")
else:
    print("  WARNING: No reference mRNA Accession (`mrna_acc`) provided.")


# --- 4. Define Translation Function (Revised Logic) ---

def score_kozak_context(rna_seq, cand_idx, wt_kozak_seq, wt_anchor_idx, original_wt_idx):
    """Scores a candidate AUG based on similarity to the WT Kozak sequence."""
    base_m3 = rna_seq[cand_idx - 3] if cand_idx >= 3 else 'N'
    base_p4 = rna_seq[cand_idx + 3] if cand_idx + 3 < len(rna_seq) else 'N'
    kozak_label = "Weak"
    if (base_m3 in 'GA') and (base_p4 == 'G'): kozak_label = "Strong"
    elif (base_m3 in 'GA') or (base_p4 == 'G'): kozak_label = "Moderate"

    score, max_score = 0, 0
    cand_kozak_seq = rna_seq[max(0, cand_idx - WT_KOZAK_WINDOW_NT) : cand_idx + 3 + WT_KOZAK_WINDOW_NT]
    cand_anchor_idx = min(cand_idx, WT_KOZAK_WINDOW_NT)

    for i in range(-WT_KOZAK_WINDOW_NT, WT_KOZAK_WINDOW_NT + 1): # Covers -9 to +9
        weight = 3 if i in [-3, 3] else 2 if i in [-2, -1, 1, 2] else 1
        wt_char_idx = wt_anchor_idx + i
        cand_char_idx = cand_anchor_idx + i

        if 0 <= wt_char_idx < len(wt_kozak_seq) and 0 <= cand_char_idx < len(cand_kozak_seq):
            max_score += weight
            if wt_kozak_seq[wt_char_idx] == cand_kozak_seq[cand_char_idx]:
                score += weight

    norm_score = score / max_score if max_score > 0 else 0
    if kozak_label == "Strong": norm_score += 0.03
    elif kozak_label == "Moderate": norm_score += 0.01

    dist = cand_idx - original_wt_idx
    penalty = (abs(dist) * 0.002) if dist < 0 else (abs(dist) * 0.001)
    final_score = norm_score - penalty

    return {"idx": cand_idx, "score": final_score, "kozak": kozak_label, "m3": base_m3, "p4": base_p4, "dist": dist}


def translate_mrna_strict_aug(mrna_sequence, cds_start_index=0, find_alternative_aug=True):
    """Translates mRNA by finding the best biological start codon (AUG) using a phased search."""
    global aug_search_activations_count
    if not isinstance(mrna_sequence, str) or not mrna_sequence: return "Error: Invalid mRNA sequence"

    mrna_upper = mrna_sequence.upper().replace('T', 'U')
    seq_len = len(mrna_upper)
    effective_cds_start = None

    if cds_start_index + 3 <= seq_len and mrna_upper[cds_start_index : cds_start_index + 3] == 'AUG':
        effective_cds_start = cds_start_index
    elif not find_alternative_aug:
        return f"Error: No AUG at index {cds_start_index} (search disabled)."

    if effective_cds_start is None:
        aug_search_activations_count += 1

        # Phase A - Codon-step pre-pass for Strong Kozak sites
        for offset in [-15, -12, -9, -6, -3, 3, 6, 9, 12, 15]:
            probe_idx = cds_start_index + offset
            if 0 <= probe_idx < seq_len - 3 and mrna_upper[probe_idx:probe_idx+3] == 'AUG':
                score_info = score_kozak_context(mrna_upper, probe_idx, wt_kozak_sequence, wt_start_in_window, cds_start_index)
                if score_info['kozak'] == "Strong":
                    effective_cds_start = probe_idx
                    print(f"  -> Pre-pass found Strong Kozak: AUG@{score_info['idx']} | -3={score_info['m3']} +4={score_info['p4']}")
                    break
        
        # Phase B: Main similarity search if pre-pass fails
        if effective_cds_start is None:
            candidates = []
            band_nt = 90
            search_region_start = max(0, cds_start_index - band_nt)
            search_region_end = min(seq_len, cds_start_index + band_nt + 3)
            
            for match in re.finditer('AUG', mrna_upper[search_region_start:search_region_end]):
                abs_idx = search_region_start + match.start()
                if abs_idx >= 3 and abs_idx + 4 < seq_len:
                    candidates.append(score_kozak_context(mrna_upper, abs_idx, wt_kozak_sequence, wt_start_in_window, cds_start_index))
            
            if candidates:
                candidates.sort(key=lambda x: x['score'], reverse=True)
                print("  -> Top candidates:")
                for cand in candidates[:3]:
                    print(f"     AUG@{cand['idx']:<5} score={cand['score']:.2f}  kozak={cand['kozak']:<8}  -3={cand['m3']} +4={cand['p4']}  dist={cand['dist']:<+4}")
                
                best_cand = candidates[0]
                second_best_score = candidates[1]['score'] if len(candidates) > 1 else 0
                if best_cand['score'] >= 0.70 or (best_cand['score'] >= 0.60 and best_cand['score'] - second_best_score >= 0.05):
                    effective_cds_start = best_cand['idx']
                    print(f"  -> Selected best context match: AUG@{best_cand['idx']}")

        # Phase C: ROBUST fallback scan if still no start found
        if effective_cds_start is None:
            search_pos = cds_start_index + 1
            weak_candidates = []
            while search_pos < seq_len:
                next_aug_idx = mrna_upper.find('AUG', search_pos)
                if next_aug_idx == -1: break # No more AUGs found
                
                if next_aug_idx >= 3 and next_aug_idx + 4 < seq_len:
                    score_info = score_kozak_context(mrna_upper, next_aug_idx, wt_kozak_sequence, wt_start_in_window, cds_start_index)
                    if score_info['kozak'] in ["Strong", "Moderate"]:
                        effective_cds_start = next_aug_idx
                        print(f"  -> Fallback scan found: AUG@{score_info['idx']} | -3={score_info['m3']} +4={score_info['p4']} | Kozak={score_info['kozak']}")
                        break # Found a good one, stop scanning
                    else:
                        weak_candidates.append(score_info) # It's weak, save for later
                
                search_pos = next_aug_idx + 1 # Continue search after this weak AUG
            
            # Optional Safety Net: If we only found weak AUGs, pick the best one
            if effective_cds_start is None and weak_candidates:
                weak_candidates.sort(key=lambda x: x['score'], reverse=True)
                best_weak = weak_candidates[0]
                effective_cds_start = best_weak['idx']
                print(f"  -> Safety net: No moderate/strong AUG found. Using best weak candidate: AUG@{best_weak['idx']}")


    if effective_cds_start is None: return "Error: No suitable AUG start codon found after search."
    try:
        protein_seq = str(Seq(mrna_upper[effective_cds_start:]).translate(to_stop=True))
        return protein_seq
    except Exception as e:
        return f"Error: Translation failed ({e})"


# --- 5. Apply Translation and Calculate Counts ---
if mrna_df is not None and cds_start is not None:
    print(f"\n--- Translating sequences (Biological AUG context search) ---")
    mrna_df['protein_sequence'] = mrna_df['generated_mRNA'].apply(
        lambda seq: translate_mrna_strict_aug(
            seq,
            cds_start_index=cds_start,
            find_alternative_aug=True
        )
    )
    print("\nTranslation attempt complete.")

    successful_translations = mrna_df[~mrna_df['protein_sequence'].str.startswith("Error:", na=True)]
    early_stop_count = successful_translations['protein_sequence'].str.contains('\*', na=False, regex=False).sum()
    successful_count = len(successful_translations)
    error_count = len(mrna_df) - successful_count

    print("\n--- Translation Summary ---")
    print(f"Total sequences processed: {len(mrna_df)}")
    print(f"Successfully translated (found AUG): {successful_count}")
    print(f"Failed translation (e.g., no AUG found): {error_count}")
    print(f"AUG search function activated: {aug_search_activations_count} times")
    print(f"Translations resulting in early stop codons ('*'): {early_stop_count}")

    # --- NEW: Inspect Failing Rows ---
    if error_count > 0:
        print("\n--- Inspecting Failing Rows ---")
        failing_rows = mrna_df[mrna_df['protein_sequence'].str.startswith("Error:", na=True)]
        with pd.option_context('display.max_colwidth', 150):
            print(failing_rows[['generated_mRNA', 'protein_sequence']].head())
    
    # --- 6. Save Updated DataFrame ---
    output_protein_csv_path = "/home/youss/DNA-bert/SLC19A3_SNP_PROTEIN_SEQUENCES_BiologicalAUG_v4.csv"
    try:
        output_dir = os.path.dirname(output_protein_csv_path)
        if output_dir:
            os.makedirs(output_dir, exist_ok=True)
        mrna_df.to_csv(output_protein_csv_path, index=False)
        print(f"\nSuccessfully saved the DataFrame to: {output_protein_csv_path}")
    except Exception as e:
        print(f"\nERROR saving DataFrame to CSV: {e}")

else:
    print("\n--- Skipping translation: input DataFrame or reference CDS info is missing. ---")

print("\n--- End of Script ---")
```

```python
def check_protein_length_consistency(dataset):
    """
    Checks for consistency in protein lengths for rows with 'normal' clinical significance.
    If a mismatch is found, it prints the row index and the mismatched length.

    Args:
        dataset (pd.DataFrame): The dataset containing the protein sequences and clinical significance.

    Returns:
        tuple: (expected_protein_length, found_inconsistency)
    """
    found_first = False
    found_inconsistency = False
    protein_length = 0

    for i, row in dataset.iterrows():
        if row['clinical_significance'] == 'normal':
            current_length = len(row['protein_sequence'])
            if not found_first:
                # Set the reference protein length
                found_first = True
                protein_length = current_length
                print(f"Row {i}: Protein length = {protein_length}")
            elif current_length != protein_length:
                # Report inconsistency
                found_inconsistency = True
                print(f"Row {i}: Protein length mismatch! Expected {protein_length}, got {current_length}")

    return protein_length, found_inconsistency


# Call the function with the dataset
expected_length, inconsistency_found = check_protein_length_consistency(final_dataset)

# Summary
if inconsistency_found:
    print("\nInconsistencies in protein lengths were found.")
else:
    print("\nAll protein lengths for 'normal' clinical significance are consistent.")
print(f"Expected protein length: {expected_length}")
```

```python
def analyze_stop_codons(final_dataset, expected_length):
    """
    Analyze rows in the dataset for stop codons in the function_class column and their impact on protein length.
    Also fixes clinical significance for variants with normal protein length but pathogenic labels.

    Args:
        final_dataset (pd.DataFrame): The dataset containing the data to analyze.
        expected_length (int): The expected protein length for normal cases.

    Returns:
        dict: A dictionary containing the counts of stop codon-related labels and their classifications.
    """
    stop_codons = ['UAG', 'UAA', 'UGA']
    number_of_stop_labels = 0
    number_of_stop_labels_and_abnormal = 0
    number_of_stop_labels_and_normal = 0
    number_of_stop_labels_pathogenic_despite_normal_length = 0
    rows_corrected = 0

    for i, row in final_dataset.iterrows():
        function_class = row['function_class']
        
        # Handle NaN values in function_class
        if pd.isna(function_class):
            continue
            
        if row['clinical_significance'] != 'normal':
            if 'stop' in str(function_class).lower():
                len_protein = len(row['protein_sequence'])
                number_of_stop_labels += 1
                if len_protein != expected_length:
                    number_of_stop_labels_and_abnormal += 1
                elif len_protein == expected_length:
                    number_of_stop_labels_and_normal += 1
                    print(f"Row {i}: Protein length is normal ({len_protein}) but has a stop codon in function_class: {function_class}")
                    print(f"SNP ID of row {i} is {row['snp_id']}, Mutant: {row['mutant']}")
                    if 'pathogenic' in str(row['clinical_significance']).lower():
                        print(f"Row {i} has pathogenic clinical significance despite normal protein length.")
                        print(f"Changing clinical significance from '{row['clinical_significance']}' to 'likely-benign'")
                        number_of_stop_labels_pathogenic_despite_normal_length += 1
                        # Fix the clinical significance
                        final_dataset.at[i, 'clinical_significance'] = 'likely-benign'
                        rows_corrected += 1

    results = {
        'number_of_stop_labels': number_of_stop_labels,
        'number_of_stop_labels_and_abnormal': number_of_stop_labels_and_abnormal,
        'number_of_stop_labels_and_normal': number_of_stop_labels_and_normal,
        'number_of_stop_labels_pathogenic_despite_normal_length': number_of_stop_labels_pathogenic_despite_normal_length,
        'rows_corrected': rows_corrected
    }

    return results

# Run the analysis and correction
results = analyze_stop_codons(final_dataset, expected_length)
print(results)
print(f"\nCorrected {results['rows_corrected']} rows with pathogenic labels but normal protein length.")
```
