```python
def calculate_label_statistics(df, label="Pathogenic"):
    
    filtered_df = df[df['Predicted_Label_Display'] == label]
    number_of_variants = filtered_df.shape[0]
    number_of_mutated_protien_seq = filtered_df[filtered_df['normal_seq'] == False].shape[0]
    number_of_normal_protien_seq = filtered_df[filtered_df['normal_seq'] == True].shape[0]
    mutated_percentage_to_total = (number_of_mutated_protien_seq / number_of_variants * 100) if number_of_variants > 0 else 0
    average_confidence = filtered_df['Prediction_Confidence'].mean()
    average_confidence_normal = filtered_df[filtered_df['normal_seq'] == True]['Prediction_Confidence'].mean()
    average_confidence_mutated = filtered_df[filtered_df['normal_seq'] == False]['Prediction_Confidence'].mean()

    return {
        "number_of_variants": number_of_variants,
        "number_of_mutated_protien_seq": number_of_mutated_protien_seq,
        "number_of_normal_protien_seq": number_of_normal_protien_seq,
        "mutated_percentage_to_total": mutated_percentage_to_total,
        "average_confidence": average_confidence,
        "average_confidence_normal": average_confidence_normal,
        "average_confidence_mutated": average_confidence_mutated
    }


pathogenic_old_model_stats = calculate_label_statistics(old_model, "Pathogenic")
pathogenic_new_model_stats = calculate_label_statistics(new_model, "Pathogenic")
pathogenic_newest_model_stats = calculate_label_statistics(newest_model, "Pathogenic")
pathogenic_model_4_stats = calculate_label_statistics(model_4, "Pathogenic")
pathogenic_model_5_stats = calculate_label_statistics(model_5, "Pathogenic")
pathogenic_model_6_stats = calculate_label_statistics(model_6, "Pathogenic")

likely_pathogenic_old_model_stats = calculate_label_statistics(old_model, "Likely Pathogenic")
likely_pathogenic_new_model_stats = calculate_label_statistics(new_model, "Likely Pathogenic")
likely_pathogenic_newest_model_stats = calculate_label_statistics(newest_model, "Likely Pathogenic")
likely_pathogenic_model_4_stats = calculate_label_statistics(model_4, "Likely Pathogenic")
likely_pathogenic_model_5_stats = calculate_label_statistics(model_5, "Likely Pathogenic")
likely_pathogenic_model_6_stats = calculate_label_statistics(model_6, "Likely Pathogenic")

benign_old_model_stats = calculate_label_statistics(old_model, "Benign")
benign_new_model_stats = calculate_label_statistics(new_model, "Benign")
benign_newest_model_stats = calculate_label_statistics(newest_model, "Benign")
benign_model_4_stats = calculate_label_statistics(model_4, "Benign")
benign_model_5_stats = calculate_label_statistics(model_5, "Benign")
benign_model_6_stats = calculate_label_statistics(model_6, "Benign")

likely_benign_old_model_stats = calculate_label_statistics(old_model, "Likely Benign")
likely_benign_new_model_stats = calculate_label_statistics(new_model, "Likely Benign")
likely_benign_newest_model_stats = calculate_label_statistics(newest_model, "Likely Benign")
likely_benign_model_4_stats = calculate_label_statistics(model_4, "Likely Benign")
likely_benign_model_5_stats = calculate_label_statistics(model_5, "Likely Benign")
likely_benign_model_6_stats = calculate_label_statistics(model_6, "Likely Benign")

print("pathogenic old model stats")
for stat in pathogenic_old_model_stats:
    print(f"Old Model - {stat}: {pathogenic_old_model_stats[stat]}")
print("=" * 50)

print("pathogenic new model stats")
for stat in pathogenic_new_model_stats:
    print(f"New Model - {stat}: {pathogenic_new_model_stats[stat]}")
print("=" * 50)

print("pathogenic newest model stats")
for stat in pathogenic_newest_model_stats:
    print(f"Newest Model - {stat}: {pathogenic_newest_model_stats[stat]}")
print("=" * 50)

print("Pathogenic Model 4 Stats")
for stat in pathogenic_model_4_stats:
    print(f"Model 4 - {stat}: {pathogenic_model_4_stats[stat]}")
print("=" * 50)

print("Pathogenic Model 5 Stats")
for stat in pathogenic_model_5_stats:
    print(f"Model 5 - {stat}: {pathogenic_model_5_stats[stat]}")
print("=" * 50)

print("pathogenic model 6 stats")
for stat in pathogenic_model_6_stats:
    print(f"Model 6 - {stat}: {pathogenic_model_6_stats[stat]}")
print("=" * 50)

# ===================================================

print("likely pathogenic old model stats")
for stat in likely_pathogenic_old_model_stats:
    print(f"Likely Pathogenic Old Model - {stat}: {likely_pathogenic_old_model_stats[stat]}")
print("=" * 50)

print("likely pathogenic new model stats")
for stat in likely_pathogenic_new_model_stats:
    print(f"Likely Pathogenic New Model - {stat}: {likely_pathogenic_new_model_stats[stat]}")

print("=" * 50)


print("likely pathogenic newest model stats")
for stat in likely_pathogenic_newest_model_stats:
    print(f"Likely Pathogenic Newest Model - {stat}: {likely_pathogenic_newest_model_stats[stat]}")

print("=" * 50)

print("Likely Pathogenic Model 4 Stats")
for stat in likely_pathogenic_model_4_stats:
    print(f"Model 4 - {stat}: {likely_pathogenic_model_4_stats[stat]}")
print("=" * 50)

print("Likely Pathogenic Model 5 Stats")
for stat in likely_pathogenic_model_5_stats:
    print(f"Model 5 - {stat}: {likely_pathogenic_model_5_stats[stat]}")
print("=" * 50)

print("Likely Pathogenic Model 6 Stats")
for stat in likely_pathogenic_model_6_stats:
    print(f"Model 6 - {stat}: {likely_pathogenic_model_6_stats[stat]}")
print("=" * 50)

# ===================================================

print("benign old model stats")
for stat in benign_old_model_stats:
    print(f"Benign Old Model - {stat}: {benign_old_model_stats[stat]}")
print("=" * 50)

print("benign new model stats")
for stat in benign_new_model_stats:
    print(f"Benign New Model - {stat}: {benign_new_model_stats[stat]}")
print("=" * 50)

print("benign newest model stats")
for stat in benign_newest_model_stats:
    print(f"Benign Newest Model - {stat}: {benign_newest_model_stats[stat]}")
print("=" * 50)

print("benign model 4 stats")
for stat in benign_model_4_stats:
    print(f"Model 4 - {stat}: {benign_model_4_stats[stat]}")
print("=" * 50)

print("benign model 5 stats")
for stat in benign_model_5_stats:
    print(f"Model 5 - {stat}: {benign_model_5_stats[stat]}")
print("=" * 50)

print("benign model 6 stats")
for stat in benign_model_6_stats:
    print(f"Model 6 - {stat}: {benign_model_6_stats[stat]}")
print("=" * 50)

# =================================


print("likely benign old model stats")
for stat in likely_benign_old_model_stats:
    print(f"Likely Benign Old Model - {stat}: {likely_benign_old_model_stats[stat]}")
print("=" * 50)

print("likely benign new model stats")
for stat in likely_benign_new_model_stats:
    print(f"Likely Benign New Model - {stat}: {likely_benign_new_model_stats[stat]}")
print("=" * 50)

print("likely benign newest model stats")
for stat in likely_benign_newest_model_stats:
    print(f"Likely Benign Newest Model - {stat}: {likely_benign_newest_model_stats[stat]}")
print("=" * 50)

print("Likely Benign Model 4 Stats")
for stat in likely_benign_model_4_stats:
    print(f"Model 4 - {stat}: {likely_benign_model_4_stats[stat]}")
print("=" * 50)

print("Likely Benign Model 5 Stats")
for stat in likely_benign_model_5_stats:
    print(f"Model 5 - {stat}: {likely_benign_model_5_stats[stat]}")
print("=" * 50)

print("Likely Benign Model 6 Stats")
for stat in likely_benign_model_6_stats:
    print(f"Model 6 - {stat}: {likely_benign_model_6_stats[stat]}")
print("=" * 50)
```

```python
import re
import os
import time
import pandas as pd
from Bio.Seq import Seq
from Bio import Entrez, SeqIO
from warnings import filterwarnings

from Bio.Seq import Seq

def translate_mrna(mrna_seq, start_pos=178):
    """Translate mRNA to protein sequence using BioPython"""
    
    if mrna_seq[start_pos:start_pos+3] != 'AUG':
        raise ValueError("Start codon not found at the specified position")
    else:
        print("Start codon found")
    
    # Extract the coding sequence from start position to end (or 1088)
    coding_seq = mrna_seq[start_pos:1672]
    
    # Use BioPython's translate method - it stops at the first stop codon automatically
    protein = str(Seq(coding_seq).translate(to_stop=True))
    
    return protein

# Translate the normal mRNA sequence
normal_protein = translate_mrna(normal_mrna_slc19a2)
print(f"Normal protein sequence length: {len(normal_protein)}")

# Translate the mutated mRNA sequence
mutated_protein = translate_mrna(mrna)
print(f"Mutated protein sequence length: {len(mutated_protein)}")


if normal_protein != mutated_protein:
    print("The sequences are different!")
    # Find first difference
    for i, (n, m) in enumerate(zip(normal_protein, mutated_protein)):
        if n != m:
            print(f"First difference at position {i}: {n} -> {m}")
            break
else:
    print("The sequences are identical")
```
